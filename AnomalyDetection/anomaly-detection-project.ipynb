## 5. Anomaly Detection using IQR Method

```python
# Step 5. Anomaly detection using IQR method for every feature
dfIQR = uniquedf.copy()
# We define functions for this step.
def find_outliers_IQR(mydf, column_name):
  sorted = mydf[column_name].sort_values()
  # Calculate the quantiles.
  q1 = sorted.quantile(0.25)
  q3 = sorted.quantile(0.75)
  iqr = q3 - q1
  lb = q1 - 1.5 * iqr
  ub = q3 + 1.5 * iqr
  print('Column_name = ', column_name)
  print('Lower bound = ', lb)
  print('Upper bound = ', ub)
  new_colname = column_name + '_anomaly'
  mydf[new_colname] = (mydf[column_name] < lb) | (mydf[column_name] > ub)
  mydf[new_colname] = mydf[new_colname].apply(lambda x: 1 if x else 0)

  return mydf

for col in dfIQR.columns:
  df_outliersIQR = find_outliers_IQR(dfIQR, col)
#df_outliers = find_outliers_IQR(dfIQR, 'Engine rpm')
df_outliersIQR.head()
```

## 6a. Feature Scaling

Recall that the dataframe **uniquedf** has the original data for ship's engine.

```python
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler

# Mute warnings.
import warnings
warnings.filterwarnings('ignore')

X = uniquedf.copy()
# Scale data for better SVM performance
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
dfXscaled = pd.DataFrame(X_scaled, columns=X.columns)
dfXscaled.head()
```

## 6b. One-Class Supervised Vector Machine (OCSVM)

```python
# Example usage with a pre-defined One-Class SVM model
model = OneClassSVM(kernel='rbf', gamma=0.5, nu=0.05)
model.fit(X_scaled)

# Predict the class (normal or anomaly).
y_pred = model.predict(X_scaled)
#print(y_pred.shape)

ocsvmdf = pd.DataFrame(X_scaled, columns=dfXscaled.columns)
ocsvmdf['anomaly'] = y_pred

# Select two variables from 6 using PCA
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca.fit(X_scaled)
X_pca = pca.transform(X_scaled)

# Visualise the data
def plot_model(model, xscaled, xpca, gamma, nu):
    # Create a grid for plotting decision boundaries and anomaly detection.
    xx, yy = np.meshgrid(np.linspace(-3, 3, 500), np.linspace(-3, 3, 500))

    # Use the first two features of X_scaled for plotting
#    X_plot = X_scaled[:, :2]

    Z = model.decision_function(np.c_[xx.ravel(), yy.ravel(),
                                      xx.ravel(), yy.ravel(),
                                      xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    # Predict whether each point is an inlier (1) or an outlier (-1).
    y_pred = model.predict(xscaled)

    # Plot setup.
    plt.figure(figsize=(12, 8))
    title = "One-Class SVM with Anomaly Detection - gamma = " + str(gamma) + " and nu = "+str(nu)
    plt.title(title)

    # Plot the decision function levels and decision boundary.
    plt.contourf(xx, yy, Z, levels=np.linspace(Z.min(), 0, 7), cmap=plt.cm.PuBu, alpha=0.8)
    plt.contour(xx, yy, Z, levels=[0], linewidths=2, colors='darkred')  # Decision boundary
    plt.contourf(xx, yy, Z, levels=[0, Z.max()], colors='palevioletred', alpha=0.5)  # Anomaly regions

    # Plot also the data points, using different markers for normals and anomalies.
    sns.scatterplot(x=xpca[:, 0], y=xpca[:, 1], hue=np.where(y_pred == 1, 'Normal', 'Anomaly'),
                    style=np.where(y_pred == -1, 'Anomaly', 'Normal'), markers={'Anomaly': 'X', 'Normal': 'o'},
                    palette={'Normal': 'deepskyblue', 'Anomaly': 'red'}, alpha=0.6, edgecolor='k')

    # Plot aesthetics.
    plt.axis('tight')
    plt.xlim((-3, 3))
    plt.ylim((-3, 3))
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.legend()
    plt.show()
plot_model(model, X_scaled, X_pca, 0.5, 0.05)
```
