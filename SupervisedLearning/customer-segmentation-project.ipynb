## 1a. Changing Currency to Float
Three features **'Total Revenue'**, **'Unit Cost'** and **'Profit'** currently are written as currency (with the currency sign) need to be changed into floating point numbers. Thus these features will be numerical as well.

```python
def currency_to_float(x):
  if isinstance(x, str):
    x = x.replace('$', '')
    if x.find(',') != -1:
      x = x.replace(',', '')
    if ((x.find('(') != -1) & (x.find(')') != -1)):
      x = x.replace('(', '')
      x = x.replace(')', '')
  return float(x)
numericData = data.copy()
numericData['Total Revenue'] = numericData['Total Revenue'].apply(currency_to_float)
numericData['Unit Cost'] = numericData['Unit Cost'].apply(currency_to_float)
numericData['Profit'] = numericData['Profit'].apply(currency_to_float)
print(numericData.shape)
numericData.head(10)
```

# 6. Selecting Optimum Choice for Number of Clusters ($k$)

## 6a. The Elbow Method

After finding that the Hierarchical Method would cause crash on this size (68300 rows) of data, we decided to run both the Elbow and the Silhouette methods for a sample data size of 30,000, similar to the Hierarchical (Dandrogram) technique.

```python

# Import Kmeans function.
from sklearn.cluster import KMeans
small_df = analysisdf.sample(n=30000).reset_index(drop=True)
# Elbow method: try different k.
wcss=[]
for i in range(1,20):
    kmeans=KMeans(n_clusters=i, init='k-means++',random_state=0, n_init = 10)
    kmeans.fit(small_df)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,20),wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# Using our best "k" and plotting the results.
clusterer = KMeans(n_clusters = 15,
                   random_state=10,
                   n_init = 10)

cluster_labels = clusterer.fit_predict(scaledf)
print(cluster_labels)
```

The output of the Elbow method suggests that the optimum number of clusters for this data is n = 4.

## 6b. The Silhouette Method

```python
# Import necessary library.
from sklearn.metrics import silhouette_samples, silhouette_score
import matplotlib.cm as cm
import matplotlib.pyplot as plt
import numpy as np

myX = small_df.to_numpy()
range_n_clusters = [2, 3, 4, 5, 6, 7, 8]
for n_clusters in range_n_clusters:
    # Create a subplot with 1 row and 2 columns.
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot.
    # The silhouette coefficient can range from -1, 1 but in this example all
    # lie within [-0.1, 1].
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(myX) + (n_clusters + 1) * 10])

    # Initialize the clusterer with n_clusters value and a random generator
    # seed of 10 for reproducibility.
    clusterer = KMeans(n_clusters=n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(myX)

    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(myX, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(myX, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them.
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle.
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot.
        y_lower = y_upper + 10  # 10 for the 0 samples.

    ax1.set_title("The silhouette plot for the various clusters.")
    ax1.set_xlabel("The silhouette coefficient values")
    ax1.set_ylabel("Cluster label")

    # The vertical line for average silhouette score of all the values.
    ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

    ax1.set_yticks([])  # Clear the yaxis labels / ticks.
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

    # 2nd Plot showing the actual clusters formed.
    colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
    ax2.scatter(myX[:, 0], myX[:, 1], marker='.', s=30, lw=0, alpha=0.7,
                c=colors, edgecolor='k')

    # Labeling the clusters.
    centers = clusterer.cluster_centers_
    # Draw white circles at cluster centers.
    ax2.scatter(centers[:, 0], centers[:, 1], marker='o',
                c="white", alpha=1, s=200, edgecolor='k')

    for i, c in enumerate(centers):
        ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1,
                    s=50, edgecolor='k')

    ax2.set_title("The visualization of the clustered data.")
    ax2.set_xlabel("Feature space for the 1st feature")
    ax2.set_ylabel("Feature space for the 2nd feature")

    plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                  "with n_clusters = %d" % n_clusters),
                 fontsize=14, fontweight='bold')

plt.show()
```

From the above results and visualisations, it is clear that the Silhouette score is best for number of clusters = 2. This implies that the Elbow method has been somewhat better so far than Silhouette Method, but this needs to be investigated further.

```python
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

# We obtain a smaller sample of 30000 rows because of the notebook crashing
#small_df = analysisdf.sample(n=30000).reset_index(drop=True)
# Convert the DataFrame to a NumPy array
myX = small_df.to_numpy()
# Hierarchical clustering from scratch
def hierarchical_clustering(data, method='average'):
    dists = pdist(myX, metric='euclidean')
    linkage_matrix = linkage(dists, method=method)
    return linkage_matrix

# Compute the linkage matrix
linkage_matrix = hierarchical_clustering(myX)


# Number of clusters we want to extract
k = 4

# Retrieve the clusters
clusters = fcluster(linkage_matrix, k, criterion='maxclust')

# Plot the data and color by cluster
plt.scatter(myX[:, 0], myX[:, 1], c=clusters, cmap='viridis')
plt.title('Hierarchical clustering of Customer data, size = 30,000 rows.')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.xlim(145, 200)
plt.ylim(40, 145)
plt.colorbar(label='Cluster')
plt.show()


# Plot the dendrogram
plt.figure(figsize=(15, 10))
dendrogram(linkage_matrix)
plt.title('Dendrogram')
plt.xlabel('Sample index or (Cluster size)')
plt.ylabel('Distance')
plt.show()
```

# 8. K-Means Clustering

We use the sample dataset of 30000 rows to obtain K-means clustering. We made the cluster plotting pair-wise using a function plot_clusters, and also defined the range of every feature accordingly along its axis in every plot.

```python
small_df = analysisdf.sample(n=30000).reset_index(drop=True)
small_df.shape
```

```python
small_df.describe()
```

```python
# Perform k-means clustering using scikit-learn.
from sklearn.cluster import KMeans
# We obtain a smaller sample of 30000 rows because of the notebook crashing
small_df = analysisdf.sample(n=30000).reset_index(drop=True)

def plot_clusters(newdf, nclusters, col1, col2, xlim1, xlim2, ylim1, ylim2):
  kmeans = KMeans(n_clusters=nclusters, random_state=42,  n_init=10)
  newdf['cluster'] = kmeans.fit_predict(newdf[[col1, col2]])

  # Extracting centroids.
  centroids = kmeans.cluster_centers_

  # Plotting the clusters and centroids.
  plt.figure(figsize=(10, 8))
  for cluster_idx in range(4):
      cluster_data = newdf[newdf['cluster'] == cluster_idx]
      plt.scatter(cluster_data[col1],
                cluster_data[col2],
                label=f'Cluster {cluster_idx + 1}')
      plt.scatter(*centroids[cluster_idx],
                s=200, marker='X',
                edgecolor='black')

  plt.xlabel(col1)
  plt.ylabel(col2)
  plt.xlim(xlim1, xlim2)
  plt.ylim(ylim1, ylim2)
  plt.title('K-Means Clustering of ' + col1 + ' and ' + col2)
  plt.legend()
  plt.show()

  # Print final centroids.
  print(col1 + " and " + col2 + " - Final Centroids:", centroids)

plot_clusters(small_df, 4, 'CLV', 'Recency', 0, 19000, 2800, 5000)
plot_clusters(small_df, 4, 'CLV', 'Frequency', 0, 19000, 0, 130)
plot_clusters(small_df, 4, 'CLV', 'Avg_Unit_Cost', 0, 19000, 0, 1600)
plot_clusters(small_df, 4, 'CLV', 'Age', 0, 19000, 9000, 30000)
plot_clusters(small_df, 4, 'Recency', 'Frequency', 2800, 5000, 0, 130)
plot_clusters(small_df, 4, 'Recency', 'Avg_Unit_Cost', 2800, 5000, 0, 1600)
plot_clusters(small_df, 4, 'Recency', 'Age', 2800, 5000, 9000, 30000)
plot_clusters(small_df, 4, 'Avg_Unit_Cost', 'Age', 0, 1600, 9000, 30000)
```

# 9. Associating Cluster number with each Customer_ID

View the cluster number associated with each customer_ID. You can create a table or DataFrame. This can be done by using our DataFrame that contains one row per customer (analysisdf dataframe) for assigning a cluster to every Customer_ID.

We can define a function that carries out this task.

```python
small_df.head()
```

We implement this by using a subframe of the dataframe cleanxdf that contains the 'Customer ID' column along with all the five new features.

```python
from sklearn.cluster import KMeans
#
def assign_cluster(newdf, nclusters, col1, col2):
  kmeans = KMeans(n_clusters=nclusters, random_state=42,  n_init=10)
  newdf['cluster'] = kmeans.fit_predict(newdf[[col1, col2]])
  df1 = pd.DataFrame()
  subcleanxdf = cleanxdf[['Customer ID', col1, col2]]
  subcleanxdf = subcleanxdf.drop_duplicates(keep=False)
  df1 = subcleanxdf.merge(newdf, left_on=[col1, col2], right_on=[col1,col2], how='left')
  #df1 = df1.drop_duplicates(keep=False)
  df1 = df1[['Customer ID', 'cluster']].dropna().drop_duplicates(keep=False)
  df1 = df1.astype({'cluster': 'int32'})
  df1 = df1.reset_index(drop=True)
  return df1
IDClusterdf = assign_cluster(small_df, 4, 'CLV', 'Frequency')
# Add one to the values of the 'cluster' column
IDClusterdf['cluster'] = IDClusterdf['cluster'] + 1
IDClusterdf.head()
```

# 11. Dimension reduction with PCA and t-SNE

We Perform dimension reduction with Principal Component Analysis (PCA) and t-test Stochastic Neighbourhood E (t-SNE) technique to reduce data to 2D.

## 11a. Dimensionality Reduction using the PCA

```python
# scaleddf will contain scaled features from analysisdf
y= Mergedf['cluster']
scaledf = Mergedf.drop(['Customer ID', 'cluster'], axis=1)
#
from sklearn.preprocessing import StandardScaler
standard_scaler = StandardScaler()
#
for cols in scaledf.columns:
  scaledf[cols] = standard_scaler.fit_transform(scaledf[[cols]])
print(scaledf.shape)
scaledf.head()
```

```python
from sklearn.decomposition import PCA
# 2. Apply PCA.
pca = PCA(n_components=2)
x_pca = pca.fit_transform(scaledf)
X_pca = pd.DataFrame(x_pca, columns=['PC1', 'PC2'])
np.issubdtype(y.dtype, np.number)
# Visualise the reduced data in 2D.
def visualise_pca2D(X, title, target):
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=target)
    # Add legend to the graph
    plt.legend(handles=scatter.legend_elements()[0],
               labels=['Cluster 1', 'Cluster 2', 'Cluster 3', 'Cluster 4'])
    plt.xlabel('Component 1')
    plt.ylabel('Component 2')
    #plt.legend(loc="upper right")
    plt.title(title)
    plt.grid(True)
    plt.show()

visualise_pca2D(X_pca, "2D Visualisation of Customer Data using PCA, Sample data size = 30000", y)
```

## 11b. Dimensionality Reduction using t-SNE

```python
from sklearn.manifold import TSNE
# Transform the data with t-SNE.
TSNE_model = TSNE(n_components=2, perplexity=30.0)
TSNE_transformed_data = TSNE_model.fit_transform(scaledf)
X_tsne = pd.DataFrame(TSNE_transformed_data, columns=['PC1', 'PC2'])
visualise_pca2D(X_tsne,
        "2D Visualisation of Customer Data, using t-SNE, after Clustering", y)
```

The outputs of PCA and t-SNE are interesting in the sense that the t-SNE has shown more opening within clusters as compared to the PCA. This is, in our view, because the PCA calculates and considers the variance between the points within a cluster, hence there is little gap within a cluster.
